# Self-Hosted Qwen LLM - Interactive Notebook

This notebook demonstrates how to use the self-hosted Qwen LLM from Jupyter notebooks or Claude Code.

   ```bash
   python serve_api.py --model Qwen/Qwen2-1.5B-Instruct
   ```

2. Install required packages:
   ```bash
   pip install requests jupyter
   ```

## Setup

````python
import sys
import os

# Add parent directory to path
sys.path.insert(0, os.path.dirname(os.getcwd()))

from client import QwenClient

# Create client
client = QwenClient(base_url="http://localhost:8000")

# Check server health
health = client.health_check()
print("Server status:", health)
````

## Example 1: Simple Text Completion

````python
prompt = "Explain what a REST API is in simple terms."
response = client.complete(prompt, max_tokens=200)
print(response)
````

## Example 2: Chat Conversation

````python
messages = [
    {"role": "user", "content": "What is machine learning?"}
]

response = client.chat(messages)
print("Assistant:", response)

# Add response to conversation
messages.append({"role": "assistant", "content": response})

# Ask follow-up
messages.append({"role": "user", "content": "Give me a simple example."})
response = client.chat(messages)
print("\nAssistant:", response)
````

## Example 3: Code Analysis

````python
code = """
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quick_sort(left) + middle + quick_sort(right)
"""

question = "Explain how this sorting algorithm works and what its time complexity is."

response = client.analyze_code(code, question)
print(response)
````

## Example 4: Debugging Help

````python
buggy_code = """
def calculate_average(numbers):
    total = 0
    for num in numbers:
        total += num
    return total / len(numbers)

result = calculate_average([])
print(result)
"""

error_message = "ZeroDivisionError: division by zero"

response = client.debug_code(buggy_code, error_message)
print(response)
````

## Example 5: Data Analysis Integration

````python
# Simulate some data analysis
import json

data = {
    "temperature": [22, 24, 23, 25, 26, 24, 23],
    "humidity": [65, 70, 68, 72, 75, 71, 69],
    "days": ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
}

prompt = f"""
Analyze this weather data and provide insights:

{json.dumps(data, indent=2)}

Please provide:
1. Temperature trends
2. Correlation observations
3. Any notable patterns
"""

response = client.complete(prompt, max_tokens=400)
print(response)
````

## Example 6: Temperature Comparison

Compare responses with different temperature settings:

````python
prompt = "Write a tagline for an AI startup."

temperatures = [0.1, 0.7, 1.2]

for temp in temperatures:
    response = client.complete(prompt, max_tokens=50, temperature=temp)
    print(f"Temperature {temp}:")
    print(f"  {response}")
    print()
````

## Example 7: Batch Processing

````python
questions = [
    "What is Docker?",
    "What is Kubernetes?",
    "What is a microservice?"
]

for question in questions:
    response = client.complete(question, max_tokens=100, temperature=0.5)
    print(f"Q: {question}")
    print(f"A: {response}")
    print("-" * 60)
````

## Example 8: Custom System Prompt

````python
messages = [
    {
        "role": "system",
        "content": "You are a Python expert who explains concepts using simple analogies."
    },
    {
        "role": "user",
        "content": "Explain what a decorator is in Python."
    }
]

response = client.chat(messages, max_tokens=300)
print(response)
````

## Example 9: Interactive Q&A Session

Build a multi-turn conversation:

````python
# Start a conversation
conversation = []

def ask(question):
    """Helper function to ask questions"""
    conversation.append({"role": "user", "content": question})
    response = client.chat(conversation, max_tokens=200)
    conversation.append({"role": "assistant", "content": response})
    return response

# Ask multiple questions
print("Q1:", ask("What is a neural network?"))
print("\nQ2:", ask("How is it different from traditional algorithms?"))
print("\nQ3:", ask("Give me a simple use case example."))
````

## Example 10: Performance Monitoring

````python
import time

prompts = [
    "What is Python?",
    "What is JavaScript?",
    "What is Rust?"
]

times = []

for prompt in prompts:
    start = time.time()
    response = client.complete(prompt, max_tokens=50)
    elapsed = time.time() - start
    times.append(elapsed)
    print(f"Prompt: {prompt}")
    print(f"Time: {elapsed:.2f}s")
    print(f"Response: {response[:100]}...\n")

print(f"Average response time: {sum(times)/len(times):.2f}s")
````
